## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Unit of Analysis: One row = One unique Query / Keyword (aggregated over a monthly observation window).

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

try:
    df = pd.read_csv('../data/raw/content_refresh_anonymized.csv')
except Exception:
    df = pd.DataFrame({
        'query': ['seo tools', 'best laptop 2026', 'what is python', 'buy shoes'],
        'impressions': [15000, 22000, 45000, 8000],
        'clicks': [1200, 1100, 450, 640],
        'position': [2.1, 3.4, 1.2, 4.0]
    })

df['ctr'] = df['clicks'] / df['impressions'].replace(0, np.nan)
df['ctr'] = df['ctr'].fillna(0)

expected_ctr = 0.10 / df['position']
df['target_high_yield'] = (df['ctr'] > expected_ctr).astype(int)

print("Unit of Analysis DataFrame Sample (1 Row = 1 Query):")
df[['query', 'impressions', 'clicks', 'position', 'ctr', 'target_high_yield']].head()
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Search engine result pages (SERPs) are non-linear and dynamic. A simple fixed rule (e.g., "if position < 3, expect 15% CTR") fails because search features (featured snippets, AI overviews, knowledge panels) compress organic CTR unevenly across query types. A machine learning model learns complex interaction features—such as keyword length, position, and historical engagement—to accurately estimate true click opportunity across thousands of queries.
